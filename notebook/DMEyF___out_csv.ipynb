```python
import glob
from pathlib import Path
import pandas as pd

# Definimos el directorio base donde buscar
base_dir = Path('..\\data\\buckets\\b1\\exp\\COMPE006_2')
out_dir = Path('..\\data\\buckets\\b1\\exp\\COMPE006_2_out')
print(base_dir.exists())
# Buscamos todos los archivos 'prediccion.txt' dentro de las carpetas que empiezan con 'COMPE'
# Puedes ajustar el patrón de búsqueda según la estructura exacta de tus carpetas
patron_busqueda = base_dir / "/prediccion_*.txt"

# Listamos los archivos encontrados
archivos_predicciones = sorted(list(base_dir.glob("prediccion_*.txt")))

print(f"Se encontraron {len(archivos_predicciones)} archivos:")
for num,path in enumerate(archivos_predicciones):
    print(num+1, " ",  path)
    df = pd.read_csv(path, sep="\t")
    df_out = df.loc[df['prob'] > 0.025, 'numero_de_cliente']
    df_out.to_csv(out_dir / f"prediccion_{num+1}.csv", sep="\t", index=False, header=False)


```
